```python
def f(s, m):
    # Начальное состояние
    current_states = {s}
    
    # Выполняем указанное количество ходов
    for _ in range(m):
        stat = set()
        # Для каждого текущего состояния применяем все возможные ходы
        for stones in current_states:
            stat.add(stones + 3)  # Ход +3
            stat.add(stones + 10) # Ход +10
            stat.add(stones * 2)  # Ход *2
        current_states = stat
    
    return current_states
```

```python
def f(s, m):
    cs = {s}
    for _ in range(m):
        stat = set()
        for i in cs:
            stat.add(i+3)
            stat.add(i+10)
            stat.add(i*2)
        
        cs = stat
    return stat

print(f(1, 2))
```
